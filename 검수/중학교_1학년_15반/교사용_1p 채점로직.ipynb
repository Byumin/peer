```python
# 지명수 구하는 함수
def count_point(df, student_list, item_no) :
    import pandas as pd
    if df['psy_code'].nunique() > 1 :
        raise ValueError('psy_code 컬럼에 여러 값이 존재합니다.')
    else :
        temp_df = df[df['question_no'].isin(item_no)] # 해당 문항 필터링
        count_list = []
        for student in student_list :
            if len(temp_df[temp_df['student_code']==student]) >= 1 :
                freq = len(temp_df[temp_df['student_code']==student])
            elif len(temp_df[temp_df['student_code']==student]) == 0 :
                freq = 0
            count_list.append(freq)
        count_point_df = pd.DataFrame({'student_code': student_list, 'num_point': count_list})
        return count_point_df, df['psy_code'].iloc[0]

# 학생 리스트 구하는 함수
def student_list(df) :
    student_list = df['target_code'].unique().tolist()
    return student_list, df['psy_code'].iloc[0]
```

```python
# 공격성 규범 지수 계산
def aggression_norms(df, celling_parameter=67) :
    import pandas as pd
    student_list_obj, psy_code = student_list(df)
    aggression_point_df, psy_code = count_point(df, student_list_obj, [4,5])
    n12_point_df, psy_code = count_point(df, student_list_obj, [12])

    aggression_norms_df = pd.merge(aggression_point_df, n12_point_df, on='student_code', suffixes=('_agg', '_12no'))
    aggression_norms_df['agg*12no_num_point'] = aggression_norms_df['num_point_agg'] * aggression_norms_df['num_point_12no']
    aggression_norms_df.sort_values(by='agg*12no_num_point', ascending=False, inplace=True)
    
    top5_aggression_norms_df = aggression_norms_df.head(5)
    top5_mean = top5_aggression_norms_df['agg*12no_num_point'].mean()
    total_mean = aggression_norms_df['agg*12no_num_point'].mean()
    weighted_mean = 0.4 * (top5_mean) + 0.6 * (total_mean)
    weighted_mean = 4 * ((min(max(weighted_mean, 0), celling_parameter)-0)/celling_parameter)+1
    result_df = pd.DataFrame({'psy_code': [psy_code], 'top5_mean': [top5_mean], 'total_mean': [total_mean], 'weighted_mean': [weighted_mean]})
    result_df['reg_date'] = pd.Timestamp.now()
    result_df['purpose'] = f'"purpose=calibration;celling_parameter={celling_parameter};author=Yumin"'
    return result_df
result_df = aggression_norms(df)
print(result_df)
```

```python
def gini(count_point_df) :
    point_mean = count_point_df['num_point'].mean()
    print(f'지명수 평균 : {point_mean}')
    n = len(count_point_df)
    print(f'학생 수 : {n}')
    abs_diff_list = []
    for point in count_point_df['num_point'] :
        abs_diff = abs(point - point_mean)
        abs_diff_list.append(abs_diff)
    abs_diff_sum = sum(abs_diff_list)
    print(f'절대차 합계 : {abs_diff_sum}')
    gini = abs_diff_sum / (2 * n * point_mean)
    print(f'지위위계 Gini 계수 : {gini}')
    gini_normalization = 4 * ((min(max(gini, 0), 0.6) - 0)/ 0.6) +1
    return gini_normalization

gini_normalization = gini(count_point_df)
print(f'학급 분위기_지위위계 : {gini_normalization}')
```

```python
def gini(count_point_df) :
    point_mean = count_point_df['num_point'].mean()
    print(f'지명수 평균 : {point_mean}')
    n = len(count_point_df)
    print(f'학생 수 : {n}')
    abs_diff_list = []
    for i in count_point_df['num_point'] :
        for j in count_point_df['num_point'] :
            if i != j:
                abs_diff = abs(i - j)
                abs_diff_list.append(abs_diff)            
            else :
                pass
    abs_diff_sum = sum(abs_diff_list)
    print(f'절대차 합계 : {abs_diff_sum}')
    gini = abs_diff_sum / (2 * n * n * point_mean)
    print(f'지위위계 Gini 계수 : {gini}')
    gini_normalization = 4 * ((min(max(gini, 0), 0.6) - 0)/ 0.6) +1
    return gini_normalization

gini_normalization = gini(count_point_df)
print(f'학급 분위기_지위위계 : {gini_normalization}')
```
