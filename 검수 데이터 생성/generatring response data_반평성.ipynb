# 또래지명 파트

```python
# 학생 명단 불러오기
import pandas as pd
students = pd.read_excel('학생_명단_역할부여.xlsx')
# 컬럼 변경
students.rename(columns={'부여_역할': '역할'}, inplace=True)
students
```

```python
# 역할별 가중치 불러오기
role_weights = pd.read_excel('역할_척도_가중치_매트릭스_v2.xlsx')
# 컬럼 변경
role_weights.rename(columns={'role_name': '역할'}, inplace=True)
role_weights
```

```python
# 학생 명단과 역할별 가중치 병합
students_role_weight = pd.merge(students, role_weights, on='역할', how='left')
students_role_weight
```

```python
# 반별로 학생 번호 재부여
students_role_weight['번호'] = students_role_weight.groupby('반')['이름'].cumcount() + 1
# 번호 컬럼 앞으로 이동
cols = students_role_weight.columns.tolist()
cols.insert(1, cols.pop(cols.index('번호')))
students_role_weight = students_role_weight[cols]
students_role_weight
```

```python
# 또래지명 문항 데이터 불러오기
point_nominations = pd.read_excel('C:/Users/qndba/peer/문항 및 로직/250825_스쿨프렌즈 검사 최종 문항_저자수정요청.xlsx', sheet_name='학생문항_또래지명')
point_nominations
```

```python
# point_nominations의 scale에 해당하는 시리즈를 컬럼으로 전환
# 학생 이름과 롤을 행인덱싱으로 활용
# 틀만 우선 만들고

point_question_simulation_t = students_role_weight[['반','번호','이름','역할']].copy()
for i, new_col in enumerate(point_nominations['scale']):
    point_question_simulation_t.loc[:, f'{i}_{new_col}'] = pd.NA
point_question_simulation_t.head()
```

```python
point_question_simulation_groupby = point_question_simulation_t.groupby('반')
point_question_student_role_weight_groupby = students_role_weight.groupby('반')
```

```python
for ban, sim_df in point_question_simulation_groupby:
    # 같은 반의 role weight dataframe 가져오기
    role_weight_df = point_question_student_role_weight_groupby.get_group(ban)

    # 여기서 sim_df, role_weight_df를 함께 사용
    print(f"=== {ban}반 ===")
    print(sim_df.shape, role_weight_df.shape)

    # 예시 처리
    # result = simulate(sim_df, role_weight_df)
```

```python
import numpy as np
final_df_list = []
for class_num, sub_simulation_df in point_question_simulation_groupby:
    print(f"Processing class: {class_num}")
    students_role_weight = point_question_student_role_weight_groupby.get_group(class_num)
    
    # 지명 이름 데이터 생성
    students_name_list = list(sub_simulation_df['이름'])

    # weights_col 정재
    weights_col = point_question_simulation_t.columns[4:]
    print(weights_col)
    
    for name in students_name_list:
        students_name_list_copy = students_name_list.copy()
        print(name)
        students_name_list_copy.remove(name)  # 본인 제거
        for col in weights_col:
            # col의 숫자와 _ 제거
            col_refined = col.split('_', 1)[1]
            print(col_refined)
            weights = list(students_role_weight.loc[students_role_weight['이름'] != name, col_refined])
            n = np.random.choice([1,2,3,4,5], p=[0.3, 0.4, 0.2, 0.05, 0.05])  # 1~5명을 확률적으로 지명
            print(len(students_name_list_copy))
            print(len(np.array(weights)/sum(weights)))
            point = np.random.choice(students_name_list_copy, p=np.array(weights)/sum(weights), replace=False, size=n)
            point = ", ".join(point.tolist())
            sub_simulation_df.loc[sub_simulation_df['이름'] == name, col] = point
    final_df_list.append(sub_simulation_df)
```

```python
final_df = pd.concat(final_df_list, ignore_index=True)
```

```python
# 시뮬레이션 결과 저장
final_df.to_excel('지명 시뮬레이션 데이터.xlsx', index=False)
```

```python
self_report_question_simulation_t_groupby = self_report_question_simulation_t.groupby('반')
self_report_question_student_role_weight_groupby = students_role_weight.groupby('반')
```

```python
weights_col = self_report_question_simulation_t.columns[4:]
print(weights_col)
```

```python
students_role_weight.columns
```

```python
weight_low_self_prob = [0.4, 0.3, 0.2, 0.05, 0.05]
weight_mid_self_prob = [0.1, 0.2, 0.4, 0.2, 0.1]
weight_high_self_prob = [0.05, 0.05, 0.2, 0.3, 0.4]
```

```python
final_df_list = []
for class_num, sub_simulation_df in point_question_simulation_groupby:
    print(f"Processing class: {class_num}")
    students_role_weight = point_question_student_role_weight_groupby.get_group(class_num)
    # 지명 이름 데이터 생성
    students_name_list = list(sub_simulation_df['이름'])

    for name in students_name_list:
        print(name)
        for col in weights_col:
            # col의 숫자와 _ 제거
            col_refined = col.split('_', 1)[1]
            weight = students_role_weight.loc[students_role_weight['이름'] == name, col_refined].values[0]
            print(col_refined, weight)
            if weight == 5.0:
                score_prob = weight_high_self_prob
            elif weight == 0.6:
                score_prob = weight_mid_self_prob
            else:
                score_prob = weight_low_self_prob
            base_scores = np.array([1, 2, 3, 4, 5])
            score = np.random.choice(base_scores, p=score_prob)
            sub_simulation_df.loc[sub_simulation_df['이름'] == name, col] = score
    final_df_list.append(sub_simulation_df)
```

```python
final_df = pd.concat(final_df_list, ignore_index=True)
print(len(final_df))
```
